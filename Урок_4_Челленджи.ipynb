## Челленджи — 3 задачи для тех, кто хочет пропустить базовый урок

Это **альтернативный, повышенный по сложности трек**: для тех, кто уже знаком с основами и хочет сразу показать владение материалом.

Чтобы подтвердить знание тем урока, выполните **все три задачи**. При этом используйте **только конструкции этого урока**: `if`, `elif`, `else`, `for`, `while`, `range`, `break`, функции, параметры, `return`, арифметику и ввод/вывод. **Без списков, словарей, классов, файлов и сторонних библиотек.**

В каждом челлендже пишите программу **в отдельной ячейке**.


### Челлендж 1. Простые числа от 2 до N

**Простое число** — целое число больше 1, которое делится без остатка только на 1 и на само себя (например, 2, 3, 5, 7, 11).

**Требования:**

1. Создайте функцию **`is_plain(n)`**, возвращающую `True` или `False`.
2. Внутри функции используйте `if`, цикл `for`, оператор `%` и `return`.
3. Запросите у пользователя целое `N` (не меньше 2).
4. С помощью второго `for` выведите **все простые числа от 2 до N включительно**.
5. Посчитайте, **сколько** простых чисел вывела программа, и выведите это количество.

**Проверка:** при `N = 10` программа должна вывести числа `2`, `3`, `5`, `7` и количество `4`.

**Ограничение:** не используйте готовые библиотеки, списки или функцию `sum()`; всё решается условиями и циклами.


### Челлендж 2. Угадай число — собственная версия с лимитом

Создайте **с нуля** функцию **`start_game(secret, limit)`**.

**Требования:**

1. При вызове функции передаётся число `secret` и максимальное количество попыток `limit`.
2. Пока есть попытки, программа просит целое число через `input()`.
3. При неверном ответе даёт подсказку `Больше` или `Меньше`.
4. При верном ответе сообщает точный **номер попытки** и возвращает `True`.
5. При исчерпании лимита сообщает правильный ответ и возвращает `False`.


**Ограничение:** без `random` и списков; используйте только `def`, `while`, условия, `return`, счётчик и ввод.


### Челлендж 3. Мини-банкомат с меню

Напишите программу, которая начинает работу с баланса **10 000 тг** и показывает пользователю меню:

```text
1 — Показать баланс
2 — Пополнить
3 — Снять деньги
0 — Выйти
```

**Требования:**

1. Меню повторяется в цикле **`while`**, пока человек не выберет `0`.
2. При выборе `1` отображается текущий баланс.
3. При выборе `2` программа спрашивает сумму пополнения и увеличивает баланс **только если сумма больше нуля**.
4. При выборе `3` программа спрашивает сумму снятия: можно снять **только положительную сумму, не превышающую баланс**. Иначе нужно показать понятное сообщение об ошибке.
5. При выборе `0` цикл завершается, выводится итоговый баланс.
6. Неизвестные команды (например, `9`) обрабатываются через `else` и **не завершают** программу.
7. **Обязательно создайте как минимум две функции**, например `пополнить(баланс, сумма)` и `снять(баланс, сумма)`, которые **возвращают** новый баланс.

**Проверка:** баланс 10000 → пополнить 500 → снять 2000 → итог 8500 тг. Попытка снять 9000 после этого должна быть отклонена.

**Ограничение:** без списков, словарей и классов. Используйте `while`, `if / elif / else`, функции и `return`.
