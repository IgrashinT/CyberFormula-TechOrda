
## Домашнее задание — 3 задачи

**Для всех, кто проходит урок вместе с преподавателем.** Здесь задачи идут от простого к комбинированному.

**Правила:**
- Каждая задача выполняется **в собственной кодовой ячейке сразу под условием**.
- Используйте только изученное: переменные, `input()`, `int()`, `print()`, f-строки, `if / elif / else`, циклы, функции.
- Не нужны списки, словари, классы и сторонние библиотеки.
- Код должен работать при разных значениях, а не только при одном тесте.
- **Решения не приведены:** их нужно написать самостоятельно.


### ДЗ 1. Температура для растений — условия

**Напишите программу**, которая:

1. Спрашивает температуру в градусах (`int(input(...))`).
2. Если температура **ниже 10**, выводит `Холодно — занесите растения домой`.
3. Иначе, если температура **от 10 до 25 включительно**, выводит `Комфортная температура`.
4. Во всех остальных случаях выводит `Жарко — проверьте полив`.

**Примеры проверки:**
- Ввод: `5` → `Холодно — занесите растения домой`
- Ввод: `20` → `Комфортная температура`
- Ввод: `30` → `Жарко — проверьте полив`

**Обязательно:** используйте **`if`, `elif` и `else`**.


### ДЗ 2. Сумма нечётных чисел — цикл `for` и `range`

**Напишите программу**, которая:

1. Запрашивает положительное целое число `N`.
2. С помощью **`for` и `range(1, N + 1, 2)`** проходит по нечётным числам от 1 до N включительно.
3. Складывает эти числа в переменную `сумма`.
4. Выводит итог.

**Примеры проверки:**
- Ввод: `5` → сумма `1 + 3 + 5 = 9` → вывод: `9`
- Ввод: `8` → сумма `1 + 3 + 5 + 7 = 16` → вывод: `16`


### ДЗ 3. Стоимость заказа — функция с условием и обработкой ошибок

Создайте программу для расчёта стоимости заказа с учётом скидки и проверкой правильности введённых данных.

**Требования к программе:**

1. Создайте функцию `show_order(price, quantity)`, которая принимает цену одного товара и его количество.
2. Функция должна вычислять общую стоимость заказа.
3. Если сумма заказа **от 5000 тг включительно**, применяется скидка **10%**.
4. Если сумма меньше 5000 тг, стоимость остаётся без изменений.
5. Функция должна возвращать итоговую стоимость с помощью `return`.
6. Вне функции программа должна запрашивать цену и количество через `input()`.
7. Если пользователь введёт **текст вместо числа**, программа не должна завершаться с ошибкой. Необходимо вывести сообщение и предложить повторить ввод.
8. Если пользователь введёт отрицательную цену или количество меньше 1, программа должна попросить ввести корректное значение.
9. Используйте цикл `while`, чтобы повторять ввод до получения корректных данных.
10. Вызовите `show_order()` с введёнными значениями и выведите результат через `print()`.
11. Отдельно проверьте функцию минимум на двух примерах.

**Пример работы программы:**

```text
Введите цену товара: hello
Ошибка! Необходимо ввести число.

Введите цену товара: -100
Ошибка! Цена не может быть отрицательной.

Введите цену товара: 1000
Введите количество: abc
Ошибка! Необходимо ввести целое число.

Введите количество: 5

Стоимость без скидки: 5000 тг
Скидка: 10%
Итоговая стоимость: 4500.0 тг
```

**Примеры проверки функции:**
- `show_order(1000, 3)` → `3000`
- `show_order(1000, 5)` → `4500.0`

**Самостоятельное изучение (обязательная часть задания):**

Для выполнения задачи самостоятельно изучите:

- `try` — выполнение кода, в котором может возникнуть ошибка.
- `except` — обработка ошибок без завершения программы.
- `ValueError` — ошибка, которая может возникнуть при преобразовании текста в число.

**Обязательно использовать:** `def`, параметры, `if / else`, `return`, `input()`, `while`, `try / except`.

**Важно:** обработка ошибок является обязательной частью задачи, а не дополнительным заданием. Готовое решение необходимо написать самостоятельно.
