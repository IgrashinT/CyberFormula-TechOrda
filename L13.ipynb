### 1.1. Методы строк: `split`, `replace`, `strip`, `lower`
Из файла всегда приходит **текст** (`str`). Чтобы с ним считать, текст нужно **разрезать** и **почистить**.
Для этого у строк есть методы — их пишут через точку: `строка.метод()`.

Вспомните `orders_raw.csv` — «грязную» версию наших заказов. Вот такие строки там встречаются:
```
ORD-0001,2026-03-01,C065,Актобе,Смартфоны,Xiaomi Redmi Note 13,99000 ₸,2,оплачен
ORD-0006,2026-03-04,C091,ALMATY,Ноутбуки,ASUS VivoBook 15,319000 ₸,1, оплачен
ORD-0016,2026-03-09,C113,караганда,Аксессуары,Наушники JBL Tune 520,25 000,1,оплачен
```

| Метод | Что делает | Пример → результат |
|---|---|---|
| `s.split(",")` | режет строку по разделителю, получается **список** | `"C065,Актобе,2".split(",")` → `['C065', 'Актобе', '2']` |
| `s.split()` | без аргумента режет по пробелам | `"MacBook Air M3".split()` → `['MacBook', 'Air', 'M3']` |
| `s.replace(что, на_что)` | заменяет **все** совпадения | `"25 000".replace(" ", "")` → `'25000'` |
| `s.strip()` | убирает пробелы и перенос строки `\n` по краям | `" оплачен".strip()` → `'оплачен'` |
| `s.lower()` | делает все буквы маленькими | `"ОПЛАЧЕН".lower()` → `'оплачен'` |

````python
line = "ORD-0001,2026-03-01,C065,Актобе,Смартфоны,Xiaomi Redmi Note 13,99000 ₸,2,оплачен"

parts = line.split(",")        # режем по запятой
print(parts)                   # список из 9 строк
print(parts[3])                # город
print(parts[6])                # цена — пока это ТЕКСТ, да ещё со знаком ₸

price = parts[6].replace("₸", "").strip()   # убираем ₸, потом пробел по краям
print(price)
print(int(price) * int(parts[7]))           # цена × количество — теперь можно считать
````

````python
print(" Алматы ".strip())
print("ALMATY".lower())
print(" оплачен".strip() == "оплачен")
print("ОПЛАЧЕН".lower() == "оплачен")
print("MacBook Air M3".split())
````

> ⚠️ Метод **не меняет** строку, а возвращает новую. Чтобы сохранить результат, нужно присвоить:
> `city = city.strip()`

````python
# ПРЕДСКАЖИТЕ: изменится ли s?
s = "  Астана  "
s.strip()
print("[" + s + "]")

s = s.strip()
print("[" + s + "]")
````

### 1.2. Открываем файл: `open` и `with`
```python
with open("имя_файла", "режим", encoding="utf-8") as f:
    # здесь работаем с файлом f
```
- `with` сам закроет файл, когда блок закончится.
- `encoding="utf-8"` — чтобы русские буквы читались правильно.

| Режим | Что делает |
|---|---|
| `"r"` | **читать** (можно не писать — это режим по умолчанию) |
| `"w"` | **записать** — ⚠️ старое содержимое файла **стирается** |
| `"a"` | **дописать** в конец файла |

> В Colab созданные файлы видно слева: значок папки 📁.

### 1.3. TXT — обычный текст

````python
# Записываем
with open("cities.txt", "w", encoding="utf-8") as f:
    f.write("Алматы\n")       # \n — перенос строки
    f.write("Астана\n")
    f.write("Шымкент\n")

# Читаем весь файл
with open("cities.txt", "r", encoding="utf-8") as f:
    text = f.read()
print(text)

# Превращаем текст в список строк
cities = text.split("\n")
print(cities)                  # ПРЕДСКАЖИТЕ: откуда пустая строка в конце?
````

````python
# Читаем построчно в цикле — и чистим каждую строку
cities = []
with open("cities.txt", encoding="utf-8") as f:
    for line in f:
        cities.append(line.strip())
print(cities)
````

### 1.4. CSV — таблица в текстовом файле
CSV — это таблица, где каждая строка файла — строка таблицы, а значения разделены **запятыми**.
Первая строка — **заголовки**. Такие файлы открывают Excel и Google Таблицы.

````python
# Смотрим на файл как на обычный текст
with open("orders.csv", encoding="utf-8") as f:
    lines = f.read().strip().split("\n")

print("строк в файле:", len(lines))
print(lines[0])      # заголовки
print(lines[1])      # первый заказ

header = lines[0].split(",")
print(header)        # названия колонок списком
````

Резать каждую строку через `split(",")` можно, но если внутри значения окажется запятая, всё сломается.
Поэтому для CSV есть готовый модуль **`csv`**. `csv.DictReader` превращает каждую строку в **словарь**:
ключи — заголовки колонок.

````python
import csv

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))      # список словарей

print(len(rows))          # ПРЕДСКАЖИТЕ: столько же, сколько строк в файле?
print(rows[0])
print(rows[0]["price"])
print(type(rows[0]["price"]))           # ПРЕДСКАЖИТЕ тип
````

> Всё, что прочитано из файла, — это **строки**, даже цена. Для счёта переводим: `int(rows[0]["price"])`.

### 1.5. JSON — словари и списки в файле
JSON — формат, в котором программы обмениваются данными. Он выглядит почти как словарь Python.
- `json.dump(данные, f)` — **записать** в файл
- `json.load(f)` — **прочитать** из файла

````python
import json

report = {"магазин": "электроника", "заказов": len(rows), "колонки": header}
with open("report.json", "w", encoding="utf-8") as f:
    json.dump(report, f,  ensure_ascii=False,  indent=2)
    # ensure_ascii=False — русские буквы сохраняются как есть
    # indent=2 — каждый ключ с новой строки, отступ 2 пробела
    
with open("report.json", encoding="utf-8") as f:
    loaded = json.load(f)

print(loaded)
print(loaded["заказов"])
````

### 2.1. Зачем нужен NumPy
**NumPy** — библиотека для работы с числами. Её основа — **массив** (`array`): набор чисел одного типа.
С массивом можно делать действие **сразу со всеми числами**, без цикла `for`.
В Colab NumPy уже установлен, нужно только подключить его:

````python
import numpy as np        # np — короткое имя, так пишут все
prices = [100, 200, 300]
arr = np.array(prices)    # из списка — в массив

print(prices)
print(arr)
print(type(arr))
````

````python
# ПРЕДСКАЖИТЕ оба результата
print(prices * 2)
print(arr * 2)
````

````python
# Список так не умеет, а массив — умеет
print(arr + 50)          # прибавить ко всем
print(arr * 1.12)        # +12% ко всем ценам
print(arr.sum(), arr.mean(), arr.max(), arr.min())   # сумма, среднее, максимум, минимум
````

### 2.2. Как ещё создать массив
| Функция | Что делает | Пример → результат |
|---|---|---|
| `np.array(список)` | из списка | `np.array([5, 7])` → `[5 7]` |
| `np.arange(начало, конец, шаг)` | как `range`, конец **не входит** | `np.arange(0, 10, 2)` → `[0 2 4 6 8]` |
| `np.zeros(n)` | `n` нулей | `np.zeros(3)` → `[0. 0. 0.]` |
| `np.ones(n)` | `n` единиц | `np.ones(3)` → `[1. 1. 1.]` |
| `np.linspace(a, b, n)` | `n` чисел от `a` до `b`, конец **входит** | `np.linspace(0, 1, 5)` → `[0. 0.25 0.5 0.75 1.]` |

````python
print(np.arange(1, 13))
print(np.arange(0, 100, 10))
print(np.zeros(5))
print(np.ones(4))
print(np.linspace(0, 1, 5))
````

### 2.3. Индексация и срезы — как у списков
Номера (индексы) начинаются с **0**. `-1` — последний элемент. Срез `a[2:5]` — с 2-го по 4-й (5 не входит).

````python
a = np.arange(10, 21)
print(a)
print(a[0], a[3], a[-1])
print(a[2:5])          # с индекса 2 до 5 (не включая)
print(a[:3])           # первые три
print(a[-3:])          # последние три
````

````python
# Позиция максимума: argmax — НОМЕР, а не значение
scores = np.array([70, 95, 88, 95, 60])
print(scores.max())      # само значение
print(scores.argmax())   # ПРЕДСКАЖИТЕ: 1 или 3?
````

### 2.4. Форма массива и `reshape`
`shape` — форма массива. У массива из 12 чисел форма `(12,)` — одна строка.
`reshape(строки, столбцы)` раскладывает те же числа в **таблицу**. Числа не меняются — меняется только форма.
Главное правило: **строки × столбцы = количество чисел** (12 = 3 × 4 = 4 × 3 = 2 × 6).

````python
x = np.arange(1, 13)
print(x)
print(x.shape)

t = x.reshape(3, 4)     # 3 строки, 4 столбца
print(t)
print(t.shape)

print(x.reshape(4, 3))  # 4 строки, 3 столбца
````

````python
# ПРЕДСКАЖИТЕ: что произойдёт?
x.reshape(5, 2)
````

> Ошибка: 12 чисел нельзя разложить в таблицу 5 × 2 = 10 ячеек.

### 2.5. Индексация таблицы: `[строка, столбец]`
```
           столбец 0   1   2
строка 0         [ 1,  2,  3]
строка 1         [ 4,  5,  6]
строка 2         [ 7,  8,  9]
строка 3         [10, 11, 12]
```

````python
t = np.arange(1, 13).reshape(4, 3)
print(t)
print(t[1])         # вся строка 1
print(t[1, 2])      # строка 1, столбец 2
print(t[-1])        # последняя строка
print(t[:, 0])      # весь столбец 0  (":" = «все строки»)
print(t[0].sum())   # сумма первой строки
````

## Часть 3. Практика: `orders_1.csv` → NumPy → JSON — 25 минут
Соберём всё вместе. Директор магазина просит короткий отчёт по заказам за март–октябрь.

### Шаг 1. Читаем CSV и собираем колонки в списки

````python
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

price = []
qty = []
city = []
product = []
status = []
for r in rows:
    price.append(int(r["price"]))
    qty.append(int(r["qty"]))
    city.append(r["city"])
    product.append(r["product"])
    status.append(r["status"])

print(price[:5])
print(qty[:5])
print(city[:5])
````

### Шаг 2. Списки → массивы NumPy

````python
price = np.array(price)
qty = np.array(qty)
city = np.array(city)
product = np.array(product)
status = np.array(status)

print(price)
print(price.shape)
````

Сумма заказа = цена × количество. Со списками понадобился бы цикл, а с массивами — **одна строка**:

````python
revenue = price * qty      # умножает 1-й на 1-й, 2-й на 2-й, ... все 500 пар сразу
print(price[:5])
print(qty[:5])
````
