# 0. Справочник

Не нужно учить наизусть — возвращайтесь сюда, когда забыли, как что пишется.

### Списки `[...]` — по порядку, можно менять

| Запись | Что делает | Пример, `a = [10, 20, 30, 40]` |
|---|---|---|
| `a[0]` | первый элемент (счёт с нуля) | `10` |
| `a[-1]` | последний элемент | `40` |
| `a[1:3]` | срез: с индекса 1 до 3, **3 не входит** | `[20, 30]` |
| `a[:2]` / `a[2:]` | первые 2 / всё начиная с индекса 2 | `[10, 20]` / `[30, 40]` |
| `len(a)` | сколько элементов | `4` |
| `x in a` | есть ли элемент | `20 in a` → `True` |
| `a.append(x)` | добавить в конец | |
| `a.remove(x)` | удалить значение `x` (первое найденное) | |
| `a.sort()` | отсортировать сам список | ничего не возвращает |
| `sorted(a)` | получить новый отсортированный список | старый не меняется |

### Кортежи `(...)` — по порядку, менять нельзя

| Запись | Что делает |
|---|---|
| `p = (3, 5)` | создать кортеж |
| `p[0]` | элемент по индексу — как у списка |
| `x, y = p` | распаковка: `x = 3`, `y = 5` |

### Словари `{ключ: значение}` — поиск по названию

| Запись | Что делает | Пример, `d = {"хлеб": 200}` |
|---|---|---|
| `d["хлеб"]` | значение по ключу | `200` |
| `d.get("сыр", 0)` | значение, а если ключа нет — `0` | `0` |
| `d["сыр"] = 1800` | добавить или изменить | |
| `"хлеб" in d` | есть ли такой ключ | `True` |
| `for k, v in d.items():` | перебрать все пары | |

### Множества `{...}` — только уникальные

| Запись | Что делает | `A = {1, 2, 3}`, `B = {2, 3, 4}` |
|---|---|---|
| `set(список)` | убрать повторы | `set([1, 1, 2])` → `{1, 2}` |
| `A & B` | что есть в обоих | `{2, 3}` |
| `A \| B` | всё вместе | `{1, 2, 3, 4}` |
| `A - B` | есть в A, но нет в B | `{1}` |

### Функции

| Запись | Что делает |
|---|---|
| `def f(*args):` | принимает сколько угодно значений, внутри это кортеж |
| `def f(**kwargs):` | принимает сколько угодно `имя=значение`, внутри это словарь |
| `lambda x: x * 2` | маленькая функция в одну строку |
| `list(map(f, a))` | применить `f` к каждому элементу |
| `list(filter(f, a))` | оставить элементы, для которых `f` даёт `True` |

````python
# ▶ Запустите эту ячейку один раз в начале урока — она нужна для проверки задач
def check(name, func, expected):
    try:
        got = func()
    except Exception as e:
        print(f"❌ {name}: ошибка {type(e).__name__}: {e}")
        return
    if got == expected:
        print(f"✅ {name}")
    else:
        print(f"❌ {name}: получилось {got!r}, а нужно {expected!r}")

print("Готово!")
````

Представьте, что нужно сохранить оценки 30 студентов. Можно завести 30 переменных: `score1`, `score2`, ... — но это неудобно. Список — это **одна переменная, в которой лежит много значений по порядку**, как вагоны в поезде.

## Индексы: как достать один элемент
У каждого элемента есть номер — **индекс**. Главное, что нужно запомнить: **счёт начинается с нуля**.

```
fruits =   ["яблоко", "банан", "вишня", "груша"]
индекс:        0         1        2        3
с конца:      -4        -3       -2       -1
```

Отрицательные индексы считают с конца: `-1` — всегда последний элемент, сколько бы их ни было.

````python
fruits = ["яблоко", "банан", "вишня", "груша"]

print(fruits[0])    # первый
print(fruits[2])    # третий
print(fruits[-1])   # последний

fruits[1] = "манго" # элемент можно заменить
print(fruits)

# print(fruits[10]) # IndexError: такого индекса нет — раскомментируйте и посмотрите
````

## Срезы: как достать кусок списка
Срез `список[начало:конец]` даёт новый список из нескольких элементов. **Начало входит, конец — нет** (так же, как в `range`).

Если начало не написать — берётся с самого начала. Если не написать конец — до самого конца.

| Срез | Что даёт | Как читать |
|---|---|---|
| `a[1:3]` | элементы 1 и 2 | «с первого до третьего, не включая» |
| `a[:3]` | первые три | «от начала до третьего» |
| `a[2:]` | всё с индекса 2 | «со второго до конца» |
| `a[-2:]` | последние два | «с предпоследнего до конца» |

````python
nums = [10, 20, 30, 40, 50, 60]

print(nums[1:3])    # [20, 30]
print(nums[:3])     # [10, 20, 30]
print(nums[3:])     # [40, 50, 60]
print(nums[-2:])    # [50, 60]
print(nums[::-1])   # задом наперёд — бонус
````

## Как менять список
У списка есть **методы** — команды, которые пишутся через точку после имени списка.

- `append(x)` — положить в конец. Самая частая операция: так список постепенно наполняется.
- `remove(x)` — удалить элемент **по значению** (не по номеру!). Если таких несколько — удаляется первый. Если такого нет — ошибка.
- `sort()` — отсортировать сам список. Для чисел — по возрастанию, для строк — по алфавиту. `sort(reverse=True)` — наоборот.

````python
shopping = ["хлеб", "молоко"]

shopping.append("яйца")
print(shopping)           # ['хлеб', 'молоко', 'яйца']

shopping.remove("хлеб")
print(shopping)           # ['молоко', 'яйца']

numbers = [5, 2, 9, 1]
numbers.sort()
print(numbers)            # [1, 2, 5, 9]
numbers.sort(reverse=True)
print(numbers)            # [9, 5, 2, 1]
````

## `sort()` или `sorted()` — частая путаница
- `список.sort()` **меняет сам список** и ничего не возвращает (точнее, возвращает `None`).
- `sorted(список)` **не трогает** исходный список, а возвращает **новый**, отсортированный.

Если исходный порядок ещё понадобится — берите `sorted()`.

````python
original = [3, 1, 2]

new = sorted(original)
print(original, new)      # [3, 1, 2] [1, 2, 3] — старый не изменился

result = original.sort()
print(original, result)   # [1, 2, 3] None — вот почему нельзя писать x = x.sort()
````

## Перебор списка циклом
Цикл `for` проходит по списку и по очереди кладёт каждый элемент в переменную. Это основной способ что-то сделать со всеми элементами: посчитать, напечатать, отобрать.

````python
scores = [85, 92, 78, 64]

for s in scores:
    print("Балл:", s)

# Собрать новый список из подходящих элементов
good = []
for s in scores:
    if s >= 80:
        good.append(s)
print("Хорошие:", good)

print("Есть ли 92?", 92 in scores)   # in проверяет, есть ли элемент
````

# 3. Кортежи (7 мин)

## Что такое кортеж
Кортеж очень похож на список, но с одним отличием: **его нельзя изменить** после создания. Ни добавить, ни удалить, ни заменить элемент. Пишется в **круглых** скобках.

**Зачем, если есть список?** Есть данные, которые по смыслу не должны меняться: координаты точки, дата рождения, размер фотографии. Кортеж защищает их от случайной порчи, и читающему код сразу понятно: «это набор, который не меняется».

````python
point = (3, 5)
birthday = (15, "марта", 2005)

print(point[0])        # индексы работают как у списка
print(birthday[1:])    # и срезы тоже
print(len(birthday))

try:
    point[0] = 10      # а менять — нельзя
except TypeError as e:
    print("Ошибка:", e)

single = (7,)          # кортеж из одного элемента — нужна запятая!
print(type(single), type((7)))
````

## Распаковка
Кортеж можно «разложить» по нескольким переменным сразу. Количество переменных слева должно совпадать с количеством элементов.

Так же функция может вернуть **несколько значений**: `return a, b` на самом деле возвращает кортеж `(a, b)`, а при вызове мы его распаковываем.

````python
point = (3, 5)
x, y = point
print("x =", x, "y =", y)

def name_and_age():
    return "Тимур", 19          # вернули кортеж из двух значений

name, age = name_and_age()      # и распаковали
print(name, age)
````

# 4. Словари (15 мин)

## Что такое словарь
В списке элемент ищут по номеру: «дай третий». Но часто удобнее искать **по названию**: «сколько стоит хлеб?», «какой телефон у Алии?». Для этого есть словарь.

Словарь хранит **пары «ключ: значение»**. Как настоящий словарь: слово → его перевод. Или как контакты в телефоне: имя → номер.

Пишется в фигурных скобках: `{ключ: значение, ключ: значение}`.

````python
prices = {"хлеб": 200, "молоко": 450, "сыр": 1800}

print(prices["молоко"])     # достать значение по ключу
print(len(prices))          # сколько пар

student = {"name": "Алия", "age": 19, "city": "Астана"}
print(student["city"])
````

## Добавить, изменить, проверить
- `словарь[ключ] = значение` — если ключа не было, пара **добавится**; если был — значение **заменится**.
- `ключ in словарь` — есть ли такой ключ.
- Ключи в словаре **не повторяются**: у одного ключа одно значение.

````python
prices = {"хлеб": 200, "молоко": 450}

prices["яйца"] = 900        # добавили новый товар
prices["хлеб"] = 220        # изменили цену
print(prices)

print("сыр" in prices)      # False
print("хлеб" in prices)     # True

del prices["яйца"]          # удалить пару
print(prices)
````

## `get()` — достать без ошибки
Если попросить ключ, которого нет, через квадратные скобки, программа упадёт с ошибкой `KeyError`. Метод `get()` в этом случае спокойно вернёт `None` или запасное значение, которое вы укажете вторым аргументом.

````python
prices = {"хлеб": 200, "молоко": 450}

# print(prices["сыр"])          # KeyError — раскомментируйте и посмотрите
print(prices.get("сыр"))        # None
print(prices.get("сыр", 0))     # 0 — наше запасное значение
print(prices.get("хлеб", 0))    # 200 — ключ есть, запасное не нужно
````

## Перебор словаря
- `keys()` — все ключи
- `values()` — все значения
- `items()` — все пары; в цикле удобно сразу разложить пару на две переменные

````python
prices = {"хлеб": 200, "молоко": 450, "сыр": 1800}

print(list(prices.keys()))
print(list(prices.values()))

for product, price in prices.items():
    print(f"{product}: {price} тг")
````

## Главный приём: подсчёт с помощью словаря
Очень частая задача — посчитать, сколько раз встречается каждое значение. Словарь для этого идеален: ключ — что считаем, значение — сколько раз встретили.

Строчка `counts[x] = counts.get(x, 0) + 1` читается так: «возьми, сколько было (а если ещё не было — ноль), прибавь один и запиши обратно». Запомните её — она понадобится в практике.

````python
votes = ["пицца", "суши", "пицца", "бургер", "пицца", "суши"]

counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1
    print(v, "→", counts)        # посмотрите, как словарь растёт шаг за шагом

print("Итог:", counts)
````

# 5. Множества (8 мин)

## Что такое множество
Множество — это набор, в котором **каждое значение встречается только один раз**. Повторы автоматически выбрасываются. Порядка у множества нет, поэтому индексов тоже нет.

Самый частый способ использования — убрать повторы из списка: `set(список)`.

Внимание: пустое множество пишется `set()`, потому что `{}` — это пустой **словарь**.

````python
cities = ["Астана", "Алматы", "Астана", "Шымкент", "Астана"]
unique = set(cities)
print(unique)                 # каждый город один раз
print(len(unique))            # сколько разных городов

s = set()
s.add("кот")
s.add("кот")                  # повтор не добавится
print(s)
print("кот" in s)
````

## Операции с множествами
Множества умеют сравниваться друг с другом. Представьте два круга, которые частично перекрываются:

| Операция | Значок | Что получаем | Пример: кто ходил в кружки |
|---|---|---|---|
| пересечение | `A & B` | что есть **в обоих** | кто ходит и на шахматы, и на футбол |
| объединение | `A \| B` | **всё вместе** без повторов | все, кто ходит хоть куда-то |
| разность | `A - B` | что есть в A, но **нет** в B | кто ходит только на шахматы |

````python
chess = {"Алия", "Тимур", "Асель"}
football = {"Тимур", "Ерлан", "Асель", "Данияр"}

print("Оба кружка:", chess & football)
print("Все:", chess | football)
print("Только шахматы:", chess - football)
print("Только футбол:", football - chess)
````

# 6. Практика: какие слова встречаются чаще всего (10 мин)

Соберём всё вместе. Задача: взять текст и посчитать, сколько раз встречается каждое слово. Так работают облака тегов, подсказки в поиске, и с этого начинается любая работа компьютера с текстом.

План из трёх шагов:
1. Привести текст к маленьким буквам, чтобы «Кофе» и «кофе» были одним словом — `.lower()`.
2. Разрезать текст на список слов — `.split()`.
3. Убрать знаки препинания по краям слова — `.strip(".,!?")`.

Посмотрим на каждый шаг отдельно:

````python
text = "Кофе утром бодрит. Кофе с молоком мягче!"

step1 = text.lower()
print(step1)

step2 = step1.split()
print(step2)

step3 = []
for w in step2:
    step3.append(w.strip(".,!?"))
print(step3)
````

## `*args` — сколько угодно аргументов
Обычно функция принимает строго определённое число аргументов. А что, если хочется написать функцию, которая складывает **любое** количество чисел: два, пять, двадцать?

Для этого перед именем параметра ставят звёздочку: `*args`. Все переданные значения соберутся в **кортеж** с этим именем. Имя `args` — просто традиция, можно назвать как угодно, важна звёздочка.

````python
def show(*args):
    print("получила:", args)

show(1, 2, 3)
show("a")
show()

def total(*prices):
    result = 0
    for p in prices:
        result += p
    return result

print(total(200, 450))
print(total(200, 450, 1800, 700))
````

## `**kwargs` — сколько угодно именованных аргументов
Две звёздочки собирают аргументы вида `имя=значение` в **словарь**. Удобно, когда у функции много необязательных настроек, и неизвестно заранее, какие из них передадут.

````python
def make_order(drink, **options):
    print("Напиток:", drink)
    print("Пожелания:", options)
    for key, value in options.items():
        print(f"  {key} = {value}")

make_order("латте", size="большой", sugar=2)
make_order("чай")
````

## Область видимости: где «живёт» переменная
- Переменная, созданная **внутри функции**, — **локальная**. Она существует только пока функция работает, снаружи её не видно.
- Переменная, созданная **вне функций**, — **глобальная**. Внутри функции её можно прочитать.
- Если внутри функции присвоить переменной с таким же именем, создастся **новая локальная** переменная, а глобальная не изменится.

Правило на практику: всё, что нужно функции, передавайте **через аргументы**, а результат забирайте **через `return`**.

````python
discount = 10                     # глобальная

def price_with_discount(price):
    final = price - price * discount / 100   # discount читаем снаружи
    return final                             # final — локальная

print(price_with_discount(1000))
# print(final)                    # NameError: снаружи её не видно

def vip_price(price):
    discount = 30                 # это НОВАЯ локальная переменная
    return price - price * discount / 100

print(vip_price(1000))
print(discount)                   # глобальная осталась 10
````

## `lambda` — функция в одну строку
Иногда нужна совсем маленькая функция, буквально на одно действие. Писать для неё `def` с именем — долго. `lambda` позволяет записать её прямо на месте.

```
lambda x: x * 2
  │    │    └── что вернуть
  │    └── параметр
  └── «вот маленькая функция»
```

Эти две записи делают одно и то же:

````python
def double(x):
    return x * 2

double_lambda = lambda x: x * 2

print(double(5), double_lambda(5))
````

## `map()` — сделать одно и то же с каждым элементом
`map(функция, список)` применяет функцию к **каждому** элементу и выдаёт результаты. Это замена циклу вида «пройди по списку и каждое значение преобразуй».

Результат нужно обернуть в `list(...)`, иначе Python покажет непонятный объект `<map ...>`.

````python
prices = [200, 450, 1800]

# через цикл
with_tax = []
for p in prices:
    with_tax.append(p * 1.12)
print(with_tax)

# то же самое через map
print(list(map(lambda p: p * 1.12, prices)))

# можно передать и готовую функцию
print(list(map(str.upper, ["кот", "пёс"])))
````

## `filter()` — оставить только подходящие
`filter(функция, список)` оставляет только те элементы, для которых функция вернула `True`. Это замена циклу «пройди по списку и отбери нужные».

````python
ages = [15, 22, 17, 30, 12, 19]

# через цикл
adults = []
for a in ages:
    if a >= 18:
        adults.append(a)
print(adults)

# то же самое через filter
print(list(filter(lambda a: a >= 18, ages)))
````

## Что мы сегодня узнали

- **Список** `[ ]` — много значений по порядку; индексы с нуля, срезы `[начало:конец]`, `append`, `remove`, `sort`.
- **Кортеж** `( )` — как список, но менять нельзя; удобная распаковка `x, y = точка`.
- **Словарь** `{ключ: значение}` — поиск по названию; `get` без ошибок, `items` для перебора, приём подсчёта.
- **Множество** — только уникальные; `&`, `|`, `-` для сравнения наборов.
- **Функции**: `*args` и `**kwargs` для любого числа аргументов, локальные и глобальные переменные, `lambda`, `map` и `filter`.

Всё это — основа для работы с данными. Когда дойдём до машинного обучения, примеры, на которых учится программа, будут лежать именно в таких списках и словарях.
