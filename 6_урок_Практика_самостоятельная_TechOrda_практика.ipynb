#### ❓ Чем `print` отличается от `return`?

- `print` — **показывает** значение на экране. Дальше его использовать нельзя.
- `return` — **отдаёт** значение туда, где вызвали функцию. Его можно сохранить, сложить, сравнить.

Функция без `return` возвращает `None`.

````python
def area_print(a, b):
    print(a * b)

def area_return(a, b):
    return a * b

x = area_print(3, 4)    # на экране 12, но...
print("x =", x)         # ...в x лежит None

y = area_return(3, 4)
print("y =", y, "| y + 1 =", y + 1)
````

#### ❓ Почему `range(1, 5)` не включает 5?

Граница `stop` **никогда не входит**. Чтобы получить числа от 1 до `n` включительно, пишем `range(1, n + 1)`.

Удобно запомнить: `range(a, b)` даёт ровно `b - a` чисел.

````python
n = 5
print(list(range(1, n)))       # без n
print(list(range(1, n + 1)))   # с n
print(len(range(3, 10)))       # 10 - 3 = 7 чисел
````

#### ❓ `=` или `==`?

- `=` — **записать** значение в переменную: `x = 5`.
- `==` — **спросить**, равны ли значения: `x == 5` → `True` или `False`.

В `if` и `while` почти всегда нужно `==` (или `<`, `>`, `!=`...).

#### ❓ Когда `for`, а когда `while`?

| Знаю **сколько раз** повторять / перебираю список или строку | Знаю только, **когда остановиться** |
|---|---|
| `for` | `while` |
| «выведи 10 чисел», «пройди по каждому растению» | «пока не введут "стоп"», «пока число > 0», «пока не накопим 1000» |

Любой `for` можно переписать через `while`, но `for` короче и безопаснее: в нём не забудешь увеличить счётчик.

````python
# Одна и та же задача двумя способами
for i in range(1, 4):
    print("for:", i)

i = 1
while i <= 3:
    print("while:", i)
    i += 1    # без этой строки — бесконечный цикл
````

#### ❓ `break`, `continue`, `return` — в чём разница?

| Команда | Что делает |
|---|---|
| `continue` | пропускает **остаток текущего шага**, цикл продолжается |
| `break` | **выходит из цикла**, код после цикла выполняется |
| `return` | **выходит из всей функции** сразу (и из цикла тоже) |

````python
def demo():
    for i in range(1, 10):
        if i == 2:
            continue      # пропустили 2
        if i == 4:
            return "вышли из функции на 4"
        print(i)
    print("Эта строка не выполнится")

print(demo())
````

#### ❓ Почему `input()` возвращает строку, даже если я ввёл число?

`input()` **всегда** возвращает текст (`str`). Чтобы считать, нужно преобразовать: `int(...)` для целых, `float(...)` для дробных.

#### ❓ Как не «уронить» программу, если ввели текст вместо числа?

`try / except` — «попробуй, а если ошибка — сделай другое». Ошибка преобразования текста в число называется `ValueError`.

````python
text = "абв"   # представьте, что это ввёл пользователь
try:
    number = int(text)
    print("Число:", number)
except ValueError:
    print("Это не число!")
print("Программа продолжает работу")
````

#### ❓ Параметр и аргумент — это одно и то же?

- **Параметр** — имя в скобках при **создании**: `def water(name, litres=1):`
- **Аргумент** — значение при **вызове**: `water("Роза", 2)`

Параметры со значением по умолчанию пишутся **после** обязательных: `def f(a, b=2)` — можно, `def f(a=2, b)` — ошибка.

#### ❓ `/`, `//`, `%` — что есть что?

| Запись | Что делает | `17 ? 5` |
|---|---|---|
| `/` | обычное деление, всегда дробь | `3.4` |
| `//` | целая часть от деления | `3` |
| `%` | остаток от деления | `2` |

Полезные приёмы: `n % 2 == 0` — чётное; `n % 10` — последняя цифра; `n // 10` — число без последней цифры.

````python
n = 4725
print(n % 10)      # последняя цифра
print(n // 10)     # без последней цифры
print(n % 2 == 0)  # чётное?
````

#### 3.8. Факториал

````python
def factorial(n):
    result = 0
    i = 1
    while i <= n:
        result *= i
        i += 1
    return result

check(factorial(5), 120, "factorial(5)")
````

### Решения Части 4 «Задачи»

````python
# А1
def grade(score):
    if score < 0 or score > 100:
        return "ошибка"
    if score < 50:
        return 2
    elif score < 70:
        return 3
    elif score < 85:
        return 4
    else:
        return 5

# А2
def is_leap(year):
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

# А3
def ticket_price(age, is_student=False):
    if age < 7:
        price = 0
    elif age < 18:
        price = 500
    else:
        price = 1000
    if is_student:
        price = price // 2
    return price

check(grade(85), 5, "grade(85)")
check(is_leap(1900), False, "is_leap(1900)")
check(ticket_price(12, is_student=True), 250, "ticket_price(12, is_student=True)")
````

````python
# Б1
def sum_multiples(n, k=3):
    total = 0
    for i in range(1, n + 1):
        if i % k == 0:
            total += i
    return total

# Б2
def count_vowels(text):
    count = 0
    for letter in text:
        if letter in "аеёиоуыэюя":
            count += 1
    return count

# Б3
def multiplication_row(n):
    row = []
    for i in range(1, 11):
        row.append(n * i)
    return row

# Б4
def filter_even(numbers):
    result = []
    for n in numbers:
        if n % 2 == 0:
            result.append(n)
    return result

check(sum_multiples(20, k=4), 60, "sum_multiples(20, k=4)")
check(count_vowels("орхидея"), 4, "count_vowels('орхидея')")
check(multiplication_row(3), [3, 6, 9, 12, 15, 18, 21, 24, 27, 30], "multiplication_row(3)")
check(filter_even([-2, 0, 7]), [-2, 0], "filter_even([-2, 0, 7])")
````

````python
# В1
def reverse_number(n):
    result = 0
    while n > 0:
        result = result * 10 + n % 10
        n = n // 10
    return result

# В2
def collatz_steps(n):
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = n * 3 + 1
        steps += 1
    return steps

# ⭐ В3
def digits_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n = n // 10
    return total

def digital_root(n):
    while n >= 10:
        n = digits_sum(n)
    return n

check(reverse_number(1200), 21, "reverse_number(1200)")
check(collatz_steps(27), 111, "collatz_steps(27)")
check(digital_root(9875), 2, "digital_root(9875)")
````

````python
# Г1
def average_above(heights, limit):
    total = 0
    count = 0
    for h in heights:
        if h > limit:
            total += h
            count += 1
    if count == 0:
        return 0
    return round(total / count, 2)

# Г2
def password_strength(password):
    has_digit = False
    for symbol in password:
        if symbol in "0123456789":
            has_digit = True
    if len(password) >= 8 and has_digit:
        return "сильный"
    elif len(password) >= 6:
        return "средний"
    else:
        return "слабый"

check(average_above([10, 20, 30, 5], 15), 25.0, "average_above(...)")
check(password_strength("abcdefgh"), "средний", "'abcdefgh'")
check(password_strength("abcdefg1"), "сильный", "'abcdefg1'")
````

### Решение мини-проекта «Дневник растений»

````python
def read_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Нужно ввести целое число")


def plant_diary():
    plants = []
    while True:
        print()
        print("1 — добавить растение")
        print("2 — показать все растения")
        print("3 — найти растение")
        print("0 — выход")
        choice = read_int("Ваш выбор: ")

        if choice == 1:
            name = input("Название: ")
            if name in plants:
                print("Уже есть в списке")
            else:
                plants.append(name)
                print(f"Добавлено: {name}")
        elif choice == 2:
            if len(plants) == 0:
                print("Список пуст")
            else:
                number = 1
                for p in plants:
                    print(f"{number}. {p}")
                    number += 1
        elif choice == 3:
            name = input("Что ищем? ")
            if name in plants:
                print("Есть")
            else:
                print("Нет")
        elif choice == 0:
            print("До свидания!")
            return
        else:
            print("Нет такого пункта")

plant_diary()
````

````python
# ДЗ урока 5
def count_taller(heights, limit=20):
    count = 0
    for h in heights:
        if h > limit:
            count += 1
    return count

def find_max(numbers):
    if len(numbers) == 0:
        return None
    biggest = numbers[0]
    for n in numbers:
        if n > biggest:
            biggest = n
    return biggest

def unique(items):
    result = []
    for item in items:
        if item not in result:
            result.append(item)
    return result

def digits_list(n):
    digits = []
    while n > 0:
        digits.insert(0, n % 10)
        n = n // 10
    return digits

def is_palindrome(n):
    digits = digits_list(n)
    for i in range(len(digits) // 2):
        if digits[i] != digits[len(digits) - 1 - i]:
            return False
    return True

palindromes = []
for number in range(100, 201):
    if is_palindrome(number):
        palindromes.append(number)
print(palindromes)

check(count_taller([5, 20, 15], limit=20), 0, "count_taller(..., limit=20)")
check(find_max([-4, -1, -7]), -1, "find_max([-4, -1, -7])")
check(unique(["роза", "мак", "роза", "фикус", "мак"]), ["роза", "мак", "фикус"], "unique(растения)")
check(digits_list(4725), [4, 7, 2, 5], "digits_list(4725)")
check(is_palindrome(1231), False, "is_palindrome(1231)")
````
