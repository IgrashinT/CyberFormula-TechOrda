Каждая функция ниже **уже работает** на хороших данных, но падает на плохих. Запустите ячейку как есть и посмотрите, какая ошибка получается. Затем допишите в функцию `try / except`, чтобы она не падала.

### Задача 1. Среднее пустого списка
Если список пустой, `len(numbers)` равен нулю — получается деление на ноль. Сделайте так, чтобы для пустого списка функция возвращала `0`.

````python
def average(numbers):
    return sum(numbers) / len(numbers)

check("обычный список", lambda: average([4, 5, 6]), 5.0)
check("пустой список", lambda: average([]), 0)
````

### Задача 3. Сумма из анкеты
В анкете люди писали, сколько часов в неделю учатся, но кто-то написал словами. Функция должна сложить только то, что получилось превратить в число, а остальное **пропустить**.

Подсказка: `try / except` нужен **внутри** цикла — чтобы одна плохая строка не останавливала весь подсчёт.

````python
def sum_answers(answers):
    total = 0
    for a in answers:
        total += int(a)
    return total

check("все числа", lambda: sum_answers(["5", "10", "3"]), 18)
check("с мусором", lambda: sum_answers(["5", "много", "7", "", "три"]), 12)
````

### Задача 3. Понятная ошибка для даты
Если строка не похожа на дату, `strptime` сам бросает `ValueError`, но с непонятным английским текстом. Поймайте его и бросьте свою ошибку: `raise ValueError("Дата должна быть в формате ДД.ММ.ГГГГ")`.

Проверка `message` смотрит, какой текст у ошибки.

````python
from datetime import datetime

def parse_date(text):
    return datetime.strptime(text, "%d.%m.%Y").date()

def message(text):
    try:
        parse_date(text)
    except ValueError as e:
        return str(e)

check("правильная дата", lambda: parse_date("05.11.2026").day, 5)
check_raises("неправильная", lambda: parse_date("2026-11-05"), ValueError)
check("понятный текст", lambda: message("завтра"), "Дата должна быть в формате ДД.ММ.ГГГГ")
````
