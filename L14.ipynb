### 2.1. Векторизация: действие сразу над всем массивом
**Векторизация** — когда вместо цикла `for` по элементам мы пишем одно действие над всем массивом.
NumPy делает этот цикл сам, внутри, на языке C — поэтому это **в десятки раз быстрее**.

````python
import time

big = np.random.rand(1_000_000)       # миллион случайных чисел

t = time.time()
result = []
for x in big:                          # обычный цикл
    result.append(x * 2 + 1)
t_loop = time.time() - t

t = time.time()
result2 = big * 2 + 1                  # векторизация: одна строка
t_np = time.time() - t

print(f"цикл: {t_loop:.3f} с,  NumPy: {t_np:.4f} с,  быстрее в {t_loop / t_np:.0f} раз")
````

Что умеет векторизация:

| Запись | Что происходит |
|---|---|
| `a + b`, `a * b`, `a / b` | попарно: 1-й с 1-м, 2-й со 2-м… (длины должны совпадать) |
| `a * 0.9`, `a + 1000` | число применяется к **каждому** элементу |
| `a > 100`, `a == "Алматы"` | сравнение каждого элемента → массив `True/False` |
| `np.round(a)`, `np.sqrt(a)`, `np.abs(a)` | функция к каждому элементу |

````python
print(revenue[:5])
print(revenue[:5] * 0.9)               # скидка 10% на каждый заказ
print(np.round(revenue[:5] / 1000))    # в тысячах тенге, округлено
print(revenue[:5] > 150_000)           # крупный заказ?
````

### 2.2. Broadcasting: когда формы разные
**Broadcasting** («растягивание») — NumPy сам подгоняет меньший массив под больший, если формы совместимы.
Самый простой случай вы уже видели: `revenue * 0.9` — число «растянулось» на все 500 элементов.

Интереснее — таблица и строка. Пусть у нас 3 товара и 4 варианта скидки:

````python
prices = np.array([99000, 319000, 549000])        # 3 товара
discounts = np.array([0, 0.05, 0.10, 0.15])       # 4 скидки

col = prices.reshape(3, 1)                         # столбец: форма (3, 1)
table = col * (1 - discounts)                      # (3, 1) × (4,) → (3, 4)

print(col.shape, discounts.shape, "→", table.shape)
print(table)
````

```
 prices (3, 1)        1 - discounts (4,)              результат (3, 4)
 [[ 99000]]     ×    [1.0  0.95  0.90  0.85]   →   [[ 99000   94050   89100   84150]
 [[319000]]                                          [319000  303050  287100  271150]
 [[549000]]                                          [549000  521550  494100  466650]]
```
Столбец «растянулся» вправо на 4 скидки, строка скидок — вниз на 3 товара.

**Правило совместимости:** формы сравниваются **справа налево**; два размера совместимы, если они **равны** или один из них **1**.

| Формы | Получится? |
|---|---|
| `(500,)` и число | ✅ число растягивается |
| `(3, 1)` и `(4,)` | ✅ → `(3, 4)` |
| `(4, 5)` и `(5,)` | ✅ строка из 5 прибавится к каждой из 4 строк |
| `(4, 5)` и `(4, 1)` | ✅ столбец из 4 прибавится к каждому из 5 столбцов |
| `(3,)` и `(4,)` | ❌ `ValueError: operands could not be broadcast together` |

````python
a = np.array([100, 200, 300, 400, 500])   # форма (5,)
print(a + 1000)                            # 1000 прибавилось к каждому
````

````python
prices = np.array([99000, 319000, 549000]).reshape(3, 1)   # столбец (3, 1): 3 товара
discounts = np.array([0, 0.05, 0.10, 0.15])                # строка (4,): 4 скидки

print(prices)
print(discounts)
table = prices * (1 - discounts)
print(f"Размер: {table.shape}")
print(table)
````

## Часть 3. Статистика по осям — 12 минут
У таблицы два направления — **оси**:

```
                axis=1  →  (вдоль строки)
             нед.1  нед.2  нед.3  нед.4
 магазин A  [  10,    12,     9,    14 ]
 магазин B  [  20,    18,    25,    22 ]     axis=0  ↓  (вдоль столбца)
 магазин C  [   5,     7,     6,     8 ]
```
- `t.sum()` — сумма **всей** таблицы, одно число.
- `t.sum(axis=0)` — идём **вниз** по каждому столбцу → итог **по неделям** (4 числа).
- `t.sum(axis=1)` — идём **вправо** по каждой строке → итог **по магазинам** (3 числа).

> Подсказка: `axis` — это направление, которое **схлопывается**. `axis=0` убирает строки, остаются столбцы.

То же работает с `mean`, `max`, `min`, `argmax`, `argmin`.

````python
t = np.array([[10, 12,  9, 14],
              [20, 18, 25, 22],
              [ 5,  7,  6,  8]])

print(t.sum())               # всё
print(t.sum(axis=0))   # ПРЕДСКАЖИТЕ: по неделям
print(t.sum(axis=1))         # ПРЕДСКАЖИТЕ: по магазинам
print(t.mean(axis=1))        # средние продажи магазина в неделю
print(t.max(axis=1))      # в какую неделю у каждого магазина был максимум 
print(t.argmax(axis=1))      # в какую неделю у каждого магазина был максимум 
````

## Часть 4. Булевы маски — 15 минут
**Маска** — массив `True/False` той же длины, что данные. Получаем её сравнением, а используем, чтобы **отобрать** нужные элементы.

| Запись | Что значит |
|---|---|
| `m = status == "оплачен"` | маска: `True` там, где заказ оплачен |
| `revenue[m]` | только те элементы, где `True` |
| `m.sum()` | **сколько** `True` (True = 1, False = 0) |
| `m.mean()` | **доля** `True` |
| `(a) & (b)` | И — оба условия; **скобки обязательны** |
| `(a) \| (b)` | ИЛИ — хотя бы одно |
| `~m` | НЕ — переворачивает маску |
| `np.where(m, "да", "нет")` | новый массив: «да» там, где `True`, иначе «нет» |

````python
paid = status == "оплачен"
print(paid[:8])
print("оплачено:", paid.sum(), "из", len(paid), "— доля", paid.mean())
print("выручка оплаченных:", revenue[paid].sum())
print("отменено:", (~paid).sum())
````

````python
almaty = city == "Алматы"
print("Алматы И оплачен:", (almaty & paid).sum())
print("Алматы ИЛИ Астана:", ((city == "Алматы") | (city == "Астана")).sum())
````

## Часть 5. Итоговая практика: выручка «город × месяц» — 10 минут
Директор хочет таблицу: **строки — города, столбцы — месяцы**, в ячейке — выручка **оплаченных** заказов.
Здесь всё вместе: маски, таблица, оси.

**Шаг 1.** Месяц заказа достаём из даты `"2026-03-01"` методом `split("-")`.

````python
month = []
for r in rows:
    month.append(int(r["date"].split("-")[1]))    # "2026-03-01" → ["2026", "03", "01"] → 3
month = np.array(month)

cities = np.unique(city)
print(cities)
months = np.unique(month)
print(months)
````

**Шаг 2.** Создаём пустую таблицу нужной формы и заполняем её в двойном цикле: для каждой пары «город, месяц» — маска из трёх условий.

````python
table = np.zeros((len(cities), len(months)))     # 7 строк × 8 столбцов

for i in range(len(cities)):
    for j in range(len(months)):
        m = (city == cities[i]) & (month == months[j]) & paid
        table[i, j] = revenue[m].sum()

print(table.astype(int))
````
