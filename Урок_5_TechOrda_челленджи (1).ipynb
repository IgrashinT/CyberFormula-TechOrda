**Правила:**

- нельзя использовать `max()`, `min()`, `sum()`, `sorted()`, `.sort()`, `bin()` и готовые модули (`math` и др.);
- старайтесь писать **короткие и понятные** функции; если одна задача удобно делится на две функции, так и сделайте.

Запустите ячейку с функцией `check` первой.


```python
def check(actual, expected, label=""):
    if actual == expected:
        print(f"✅ {label}: {actual}")
    else:
        print(f"❌ {label}: получили {actual}, ожидали {expected}")
```
