Запустите ячейку с классом `Product` — это товар на складе магазина. Потом решите задачи 1–8: **каждая — одна строка кода**.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name            # название
        self.price = price          # цена за 1 шт, ₸
        self.quantity = quantity    # сколько штук на складе

    def total(self):
        return self.price * self.quantity

    def is_available(self):
        return self.quantity > 0

    def sell(self, n):
        if n > self.quantity:
            print(f"Нельзя продать {n} шт: на складе только {self.quantity}")
        else:
            self.quantity = self.quantity - n
````

13. Класс **`Temperature`**: атрибут `celsius`. Методы:
- `to_fahrenheit()` — переводит в Фаренгейты: `celsius * 9 / 5 + 32`;
- `describe()` — возвращает `"холодно"` (ниже 0), `"прохладно"` (0–15), `"тепло"` (выше 15).

**Пример проверки:**
```python
t = Temperature(25)
print(t.to_fahrenheit())   # 77.0
print(t.describe())        # тепло
print(Temperature(-10).describe())   # холодно
```

````python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return self.celsius*(9/5) + 32

    def describe(self):
        if self.celsius < 0:
            return "Холодно"
        elif self.celsius >= 0 and self.celsius <= 15:
            return "Прохладно"
        elif self.celsius > 15:
            return "Тепло"

t = Temperature(25)
print(t.to_fahrenheit())   # 77.0
print(t.describe())        # тепло
print(Temperature(-10).describe())   # холодно
````

14. Класс **`Employee`** (сотрудник): атрибуты `name`, `position`, `salary` (зарплата в месяц). Методы:
- `annual()` — зарплата за год;
- `raise_salary(percent)` — повышает зарплату на `percent` процентов (меняет атрибут);
- `info()` — возвращает строку `"Дана, аналитик: 450000 ₸"`.

**Пример проверки:**
```python
e = Employee("Дана", "аналитик", 450_000)
print(e.annual())        # 5400000
e.raise_salary(10)
print(e.info())          # Дана, аналитик: 495000.0 ₸
```

````python
class Employee:
    def __init__(self, name, position, salary):
        self.name = name
        self.position = position
        self.salary = salary

    def annual(self):
        return self.salary*12

    def raise_salary(self, percent):
        self.salary += self.salary*(percent/100)

    def info(self):
        return f"{self.name}, {self.position}: {self.salary} ₸"


e = Employee("Дана", "аналитик", 450_000)
print(e.annual())        # 5400000
e.raise_salary(10)
print(e.info())          # Дана, аналитик: 495000.0 ₸
````

15. Класс **`Student`** с оценками. Атрибуты: `name`, `grades` (в начале пустой список). Методы:
- `add_grade(grade)` — добавляет оценку, только если она от 0 до 100, иначе печатает `"Неверная оценка: 120"`;
- `average()` — средний балл (если оценок нет — `0`);
- `best()` — лучшая оценка (если оценок нет — `None`);
- `count_above(x)` — сколько оценок **больше** `x`.

**Пример проверки:**
```python
s = Student("Мадина")
print(s.average())        # 0
for g in [90, 75, 120, 88, 60]:
    s.add_grade(g)        # напечатает: Неверная оценка: 120
print(s.grades)           # [90, 75, 88, 60]
print(s.average())        # 78.25
print(s.best())           # 90
print(s.count_above(70))  # 3
```

````python
class Student:
    def __init__(self, name, grades=[]):
        self.name = name
        self.grades = grades

    def add_grade(self, grade):
        if grade > 0 and grade < 100:
            self.grades.append(grade)
        else:
            print(f"Неверная оценка: {grade}")

    def average(self):
        if self.grades:
            return sum(self.grades) / len(self.grades)
        else:
            return 0

    def best(self):
        if self.grades:
            return max(self.grades)
        else:
            return None

    def count_above(self, threshold):
        return len(list(filter(lambda x: x > threshold, self.grades)))

s = Student("Мадина")
print(s.average())        # 0
for g in [90, 75, 120, 88, 60]:
    s.add_grade(g)        # напечатает: Неверная оценка: 120
print(s.grades)           # [90, 75, 88, 60]
print(s.average())        # 78.25
print(s.best())           # 90
print(s.count_above(70))  # 3
````

17. Класс **`Order`** — один заказ.
- `__init__(self, row)` принимает **один словарь** из `rows` и раскладывает его по атрибутам: `order_id`, `date`, `customer_id`, `city`, `category`, `product`, `price`, `qty`, `status`. Цену и количество превратите в `int`;
- `total()` — сумма заказа: цена × количество;
- `is_paid()` — `True`, если статус `оплачен`;
- `month()` — месяц заказа в виде `"2026-03"` (подсказка: срез строки `date`).

После класса создайте список `orders` — по объекту `Order` на каждую строку `rows`.

````python
class Order:
    def __init__(self, row):
        self.order_id = row["order_id"]
        self.date = row["date"]
        self.customer_id = row["customer_id"]
        self.city = row["city"]
        self.category = row["category"]
        self.product = row["product"]
        self.price = int(row["price"])
        self.qty = int(row["qty"])
        self.status = row["status"]

    def total(self):
        return self.price * self.qty

    def is_paid(self):
        return self.status == "оплачен"

    def month(self):
        return '-'.join(self.date.split('-')[:2])


orders = [Order(row) for row in rows]
````
