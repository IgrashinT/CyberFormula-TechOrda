### Обработка ошибок

| Запись | Когда выполняется |
|---|---|
| `try:` | код, который **может** упасть |
| `except ValueError:` | если в `try` случилась ошибка этого типа |
| `except ValueError as e:` | то же, а в `e` лежит описание ошибки |
| `else:` | если ошибки **не было** |
| `finally:` | **всегда**, была ошибка или нет |
| `raise ValueError("текст")` | самому вызвать ошибку |

| Ошибка | Когда бывает | Пример |
|---|---|---|
| `ValueError` | значение неподходящее | `int("abc")` |
| `TypeError` | неподходящий тип | `"5" + 3` |
| `ZeroDivisionError` | деление на ноль | `10 / 0` |
| `KeyError` | нет такого ключа в словаре | `{"a": 1}["b"]` |
| `IndexError` | нет такого индекса в списке | `[1, 2][5]` |
| `NameError` | нет такой переменной | опечатка в имени |
| `FileNotFoundError` | нет такого файла | `open("нет.txt")` |

````python
# ▶ Запустите эту ячейку первой — она нужна для проверки задач
def check(name, func, expected):
    try:
        got = func()
    except Exception as e:
        print(f"❌ {name}: ошибка {type(e).__name__}: {e}")
        return
    if got == expected:
        print(f"✅ {name}")
    else:
        print(f"❌ {name}: получилось {got!r}, а нужно {expected!r}")

def check_raises(name, func, error_type):
    try:
        func()
    except error_type:
        print(f"✅ {name}")
        return
    except Exception as e:
        print(f"❌ {name}: ошибка {type(e).__name__}, а нужна {error_type.__name__}")
        return
    print(f"❌ {name}: ошибки не было, а нужна {error_type.__name__}")

print("Готово!")
````

## Напоминание
Функция — кусок кода с именем. Пишем один раз через `def`, вызываем сколько угодно. У параметра может быть **значение по умолчанию** — тогда аргумент можно не передавать. `return` отдаёт результат туда, откуда функцию вызвали, и сразу завершает её.

````python
def greet(name, greeting="Привет"):
    return f"{greeting}, {name}!"

print(greet("Алия"))                     # greeting взялся по умолчанию
print(greet("Тимур", greeting="Салем"))  # передали свой
````

## `*args` — сколько угодно аргументов
Обычно функция принимает строго определённое число аргументов. А что, если хочется написать функцию, которая складывает **любое** количество чисел: два, пять, двадцать?

Для этого перед именем параметра ставят звёздочку: `*args`. Все переданные значения соберутся в **кортеж** с этим именем. Имя `args` — просто традиция, можно назвать как угодно, важна звёздочка.

````python
def show(*args):
    print("получила:", args)

show(1, 2, 3)
show("a")
show()

def total(*prices):
    result = 0
    for p in prices:
        result += p
    return result

print(total(200, 450))
print(total(200, 450, 1800, 700))
````

## `**kwargs` — сколько угодно именованных аргументов
Две звёздочки собирают аргументы вида `имя=значение` в **словарь**. Удобно, когда у функции много необязательных настроек, и неизвестно заранее, какие из них передадут.

````python
def make_order(drink, **options):
    print("Напиток:", drink)
    print("Пожелания:", options)
    for key, value in options.items():
        print(f"  {key} = {value}")

make_order("латте", size="большой", sugar=2)
make_order("чай")
````

## Область видимости: где «живёт» переменная
- Переменная, созданная **внутри функции**, — **локальная**. Она существует только пока функция работает, снаружи её не видно.
- Переменная, созданная **вне функций**, — **глобальная**. Внутри функции её можно прочитать.
- Если внутри функции присвоить переменной с таким же именем, создастся **новая локальная** переменная, а глобальная не изменится.

Правило на практику: всё, что нужно функции, передавайте **через аргументы**, а результат забирайте **через `return`**.

````python
discount = 10                     # глобальная

def price_with_discount(price):
    final = price - price * discount / 100   # discount читаем снаружи
    return final                             # final — локальная

print(price_with_discount(1000))
# print(final)                    # NameError: снаружи её не видно

def vip_price(price):
    discount = 30                 # это НОВАЯ локальная переменная
    return price - price * discount / 100

print(vip_price(1000))
print(discount)                   # глобальная осталась 10
````

## `lambda` — функция в одну строку
Иногда нужна совсем маленькая функция, буквально на одно действие. Писать для неё `def` с именем — долго. `lambda` позволяет записать её прямо на месте.

```
lambda x: x * 2
  │    │    └── что вернуть
  │    └── параметр
  └── «вот маленькая функция»
```

Эти две записи делают одно и то же:

````python
def double(x):
    return x * 2

double_lambda = lambda x: x * 2

print(double(5), double_lambda(5))
````

## `map()` — сделать одно и то же с каждым элементом
`map(функция, список)` применяет функцию к **каждому** элементу и выдаёт результаты. Это замена циклу вида «пройди по списку и каждое значение преобразуй».

Результат нужно обернуть в `list(...)`, иначе Python покажет непонятный объект `<map ...>`.

````python
prices = [200, 450, 1800]

# через цикл
with_tax = []
for p in prices:
    with_tax.append(p * 1.12)
print(with_tax)

# то же самое через map
print(list(map(lambda p: p * 1.12, prices)))

print([p*1.12 for p in prices])  # через генератор списков

# можно передать и готовую функцию
print(list(map(str.upper, ["кот", "пёс"])))
````

## `filter()` — оставить только подходящие
`filter(функция, список)` оставляет только те элементы, для которых функция вернула `True`. Это замена циклу «пройди по списку и отбери нужные».

````python
ages = [15, 22, 17, 30, 12, 19]

# через цикл
adults = []
for a in ages:
    if a >= 18:
        adults.append(a)
print(adults)

# то же самое через filter
print(list(filter(lambda a: a >= 18, ages)))

# lambda пригождается и в sorted: сортируем пары по второму элементу
pairs = [("Тимур", 81), ("Алия", 95), ("Асель", 77)]
print(sorted(pairs, key=lambda p: p[1], reverse=True))
````

## Что такое модуль
Сам по себе Python умеет не так много: `print`, `len`, `int`, `sum`... Всё остальное разложено по **модулям** — файлам с готовыми функциями, которые кто-то уже написал и проверил. Как ящики с инструментами в мастерской: понадобился молоток — достаём нужный ящик.

Модули, которые идут вместе с Python, называются **стандартной библиотекой**: их не нужно устанавливать, только подключить. Есть и внешние модули — например, `numpy` и `pandas`, которые мы будем проходить дальше. Их ставят отдельно, но в Colab они уже установлены.

Подключают модуль командой `import`, обычно в самом начале программы.

## Три способа подключить модуль

````python
# 1. Подключить модуль целиком — перед функцией пишем имя модуля и точку
import math
print(math.sqrt(16))      # корень
print(math.pi)

# 2. Взять из модуля только нужное — дальше пишем без "math."
from math import sqrt, pi
print(sqrt(25), pi)

# 3. Подключить под коротким именем (псевдонимом)
import math as m
print(m.sqrt(9))
````

| Способ | Как вызывать | Когда удобно |
|---|---|---|
| `import math` | `math.sqrt(16)` | сразу видно, откуда функция — **самый понятный** вариант |
| `from math import sqrt` | `sqrt(16)` | функция нужна часто, а имя короткое и понятное |
| `import math as m` | `m.sqrt(16)` | у модуля длинное имя; так будет `import numpy as np`, `import pandas as pd` |

Встречается ещё `from math import *` — «взять всё». Так лучше не делать: непонятно, откуда взялось имя, и можно случайно затереть свои переменные.

Чтобы посмотреть, что лежит в модуле, есть `dir(модуль)`, а чтобы прочитать описание функции — `help(функция)`.

````python
import math
print(dir(math)[-15:])    # последние 15 имён из модуля
help(math.ceil)
````

## `math` — математика
| Функция | Что делает | Пример → результат |
|---|---|---|
| `math.sqrt(x)` | квадратный корень | `math.sqrt(49)` → `7.0` |
| `math.pi` | число π (это не функция, скобки не нужны) | `3.141592653589793` |
| `math.ceil(x)` | округлить **вверх** до целого | `math.ceil(2.1)` → `3` |
| `math.floor(x)` | округлить **вниз** до целого | `math.floor(2.9)` → `2` |
| `round(x, n)` | обычное округление (встроено, модуль не нужен) | `round(2.567, 1)` → `2.6` |

`ceil` нужен, когда «не хватает» недопустимо: 47 студентов, в автобусе 20 мест — сколько автобусов? `47 / 20 = 2.35`, но двух не хватит, нужно **3**.

````python
import math

students, seats = 47, 20
print(students / seats)              # 2.35
print(math.ceil(students / seats))   # 3 автобуса
print(math.floor(students / seats))  # 2 полностью заполненных

# Расстояние между двумя точками на плоскости — теорема Пифагора
x1, y1 = 0, 0
x2, y2 = 3, 4
print(math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2))   # 5.0
````

## `random` — случайные значения
Пригождается для игр, жеребьёвки, перемешивания данных.

| Функция | Что делает |
|---|---|
| `random.randint(a, b)` | случайное целое от `a` до `b`, **оба края включительно** |
| `random.random()` | случайное дробное от 0 до 1 |
| `random.choice(список)` | один случайный элемент |
| `random.sample(список, k)` | `k` разных случайных элементов |
| `random.shuffle(список)` | перемешать **сам** список (ничего не возвращает, как `sort`) |
| `random.seed(42)` | зафиксировать «случайность» |

Компьютерная случайность на самом деле не настоящая: числа получаются по формуле из стартового значения. Если задать это значение через `seed`, то при каждом запуске будут **те же самые** «случайные» числа. Это удобно, когда нужно, чтобы результат у всех совпадал и его можно было повторить.

````python
import random

print(random.randint(1, 6))          # бросок кубика — запустите несколько раз

students = ["Алия", "Тимур", "Асель", "Ерлан", "Данияр"]
print("К доске:", random.choice(students))
print("Дежурные:", random.sample(students, 2))

random.shuffle(students)
print("Перемешали:", students)

random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))   # с seed эти числа всегда одинаковые
````

## `datetime` — даты и время
Внутри модуля `datetime` есть два главных инструмента: `date` (только дата) и `datetime` (дата и время). Обычно их сразу достают через `from`.

Даты можно вычитать — получится промежуток, у которого есть `.days` (сколько дней).

````python
from datetime import date, datetime

today = date.today()
print("Сегодня:", today)
print(today.year, today.month, today.day)

new_year = date(today.year + 1, 1, 1)
print("До Нового года дней:", (new_year - today).days)

now = datetime.now()
print("Сейчас:", now)
````

### Дата ↔ строка
Пользователь вводит дату текстом, а показывать её хочется в привычном виде. Для этого есть два метода с **шаблоном формата**:

- `strftime` — **из даты в строку** (f = format, «оформить»);
- `strptime` — **из строки в дату** (p = parse, «разобрать»).

| Код в шаблоне | Что означает | Пример |
|---|---|---|
| `%d` | день, две цифры | `05` |
| `%m` | месяц, две цифры | `11` |
| `%Y` | год, четыре цифры | `2026` |
| `%H`, `%M` | часы, минуты | `13`, `41` |

````python
from datetime import date, datetime

d = date(2026, 12, 31)
print(d.strftime("%d.%m.%Y"))        # "31.12.2026"

text = "15.03.2005"
birthday = datetime.strptime(text, "%d.%m.%Y").date()   # .date() — оставить только дату
print(birthday, birthday.year)

print((date(2026, 11, 5) - birthday).days, "дней прожито к 5 ноября 2026")
````

## `os` — файлы и папки
Модуль `os` позволяет программе работать с файловой системой: узнать, в какой папке она запущена, что там лежит, создать папку, проверить, есть ли файл. В Colab это диск виртуального компьютера, на котором работает тетрадь (слева есть значок папки — там видно файлы).

| Функция | Что делает |
|---|---|
| `os.getcwd()` | текущая папка (current working directory) |
| `os.listdir(папка)` | список файлов и папок внутри |
| `os.makedirs("папка", exist_ok=True)` | создать папку; `exist_ok=True` — не ругаться, если уже есть |
| `os.path.exists(путь)` | есть ли такой файл или папка |
| `os.path.join("data", "a.csv")` | правильно собрать путь из частей |
| `os.path.splitext("фото.jpg")` | разделить имя и расширение: `("фото", ".jpg")` |

Почему `os.path.join`, а не просто `"data" + "/" + "a.csv"`? В Windows части пути разделяются `\`, а в Mac и Linux — `/`. `join` сам выберет правильный знак.

````python
import os

print("Я в папке:", os.getcwd())
print("Здесь лежит:", os.listdir())

os.makedirs("my_data", exist_ok=True)
print("Папка есть?", os.path.exists("my_data"))

path = os.path.join("my_data", "scores.csv")
print(path)

name, ext = os.path.splitext("report_2026.pdf")
print(name, ext)
````

## Что такое исключение
Когда Python встречает то, что выполнить невозможно — превратить `"abc"` в число, поделить на ноль, взять 10-й элемент из списка длиной 3, — он **останавливает программу** и печатает сообщение об ошибке. Такая ошибка называется **исключением** (exception).

Раскомментируйте строку в ячейке ниже и запустите, чтобы увидеть сообщение. Потом закомментируйте обратно.

````python
age = "двадцать"
# print(int(age))
````

Сообщение об ошибке (его называют **traceback**) выглядит страшно, но читать его просто: **смотрите на последнюю строку**. Там написан тип ошибки и что пошло не так:

```
ValueError: invalid literal for int() with base 10: 'двадцать'
```

«`ValueError`: неподходящее значение для `int()` — `'двадцать'`». Строки выше показывают, **где** это случилось.

### Частые ошибки
| Ошибка | Когда бывает | Пример |
|---|---|---|
| `ValueError` | тип правильный, но значение не подходит | `int("abc")` |
| `TypeError` | операция не подходит для этого типа | `"5" + 3` |
| `ZeroDivisionError` | деление на ноль | `10 / 0` |
| `KeyError` | в словаре нет такого ключа | `{"a": 1}["b"]` |
| `IndexError` | в списке нет такого индекса | `[1, 2][5]` |
| `NameError` | нет переменной с таким именем — часто опечатка | `pritn("hi")` |
| `FileNotFoundError` | нет такого файла | `open("нет.txt")` |

Ошибки в программе неизбежны: пользователь введёт буквы вместо числа, файл окажется удалён, интернет пропадёт. Хорошая программа не падает, а **обрабатывает** ошибку: объясняет, что не так, и продолжает работать.

## `try / except` — поймать ошибку
```
try:
    код, который может упасть
except ТипОшибки:
    что делать, если упал
```

Как это работает:
1. Python **пробует** выполнить блок `try`.
2. Если ошибки нет — блок `except` **пропускается**.
3. Если внутри `try` случилась ошибка — оставшиеся строки `try` **не выполняются**, Python сразу прыгает в `except`.
4. После этого программа **продолжает работать** дальше, а не падает.

````python
for text in ["42", "abc"]:
    try:
        number = int(text)
        print("Получилось число:", number)   # не выполнится, если int упал
    except ValueError:
        print(f"'{text}' — это не число")

print("Программа работает дальше")
````

### Несколько видов ошибок и `as e`
- Можно написать несколько `except` для разных ошибок — сработает тот, чей тип совпал.
- `except ValueError as e:` кладёт в переменную `e` саму ошибку; `print(e)` покажет её описание.
- Несколько типов в одном `except` пишут в скобках: `except (ValueError, TypeError):`.

````python
def divide(a, b):
    try:
        return int(a) / int(b)
    except ZeroDivisionError:
        return "На ноль делить нельзя"
    except ValueError as e:
        return f"Нужны числа ({e})"

print(divide("10", "4"))
print(divide("10", "0"))
print(divide("10", "abc"))
````

### Два важных правила
1. **Ловите конкретную ошибку.** Пустой `except:` без типа ловит вообще всё, включая ваши опечатки — программа молча сделает не то, и вы не поймёте почему.
2. **В `try` кладите только то, что может упасть.** Чем меньше блок, тем понятнее, какую ошибку вы ловите.

````python
# Плохо: опечатка в имени (NameError) спрятана — кажется, что дело в данных
try:
    result = int("5") + undefined_name
except:
    print("Что-то не так с числом")   # но дело вовсе не в числе!

# Хорошо: ловим только то, что ожидаем
try:
    result = int("5")
except ValueError:
    print("Это не число")
````

## `else` и `finally`
У `try` есть ещё два необязательных блока:

| Блок | Когда выполняется | Зачем |
|---|---|---|
| `try` | всегда первым | код, который может упасть |
| `except` | только если **была** ошибка | реакция на ошибку |
| `else` | только если ошибки **не было** | продолжение «удачного» сценария |
| `finally` | **всегда** в конце | «уборка»: закрыть файл, напечатать итог, отключиться |

Посмотрите на порядок вывода в двух случаях:

````python
def safe_divide(a, b):
    print(f"--- делим {a} на {b}")
    try:
        result = a / b
    except ZeroDivisionError:
        print("except: на ноль делить нельзя")
    else:
        print("else: результат", result)
    finally:
        print("finally: я выполняюсь всегда")

safe_divide(10, 2)
safe_divide(10, 0)
````

## `raise` — вызвать ошибку самому
Иногда Python ошибки не видит, а **мы** знаем, что значение неправильное. Например, возраст `-5` или `500` — это обычные числа, Python спокойно их примет. Чтобы функция не работала с бессмыслицей, мы сами «бросаем» ошибку:

```python
raise ValueError("понятное описание проблемы")
```

После `raise` функция сразу останавливается, как после `return`. Тот, кто вызвал функцию, может поймать эту ошибку через `try / except` — точно так же, как ошибки самого Python.

Какой тип выбрать: если значение неправильное — `ValueError`, если тип неправильный (строка вместо числа) — `TypeError`.

````python
def set_age(age):
    if age < 0 or age > 120:
        raise ValueError(f"Возраст должен быть от 0 до 120, а не {age}")
    return age

print(set_age(19))

try:
    set_age(-5)
except ValueError as e:
    print("Ошибка:", e)
````

## Шаблон «спрашивать, пока не введут правильно»
Это самый частый способ проверки ввода: бесконечный цикл, внутри `try`. Если всё в порядке — `else` выходит из цикла через `break`; если нет — объясняем ошибку и спрашиваем снова.

````python
def read_age(prompt):
    while True:
        text = input(prompt)
        try:
            age = int(text)
            if age < 0 or age > 120:
                raise ValueError("возраст должен быть от 0 до 120")
        except ValueError as e:
            print("Не подходит:", e)
        else:
            return age          # ошибки не было — возвращаем и выходим из цикла
````
