### 1. Что мы уже знаем (повторение)

| Что | Как пишется | Пример | Результат |
|---|---|---|---|
| Словарь | `{ключ: значение}` | `d = {"name": "Аружан", "score": 80}` | карточка с данными |
| Достать по ключу | `d[ключ]` | `d["name"]` | `"Аружан"` |
| Достать безопасно | `d.get(ключ, запасное)` | `d.get("age", "нет")` | `"нет"`, если ключа нет |
| Изменить значение | `d[ключ] = новое` | `d["score"] = 95` | значение заменено |
| Последний элемент списка | `список[-1]` | `[90, 75, 88][-1]` | `88` |
| Вложенный доступ | скобки «слоями» слева направо | `student["grades"][-1]` | сначала список оценок, потом последняя оценка |
| Функция | `def имя(параметры):` + `return` | `def double(x): return x * 2` | `double(5)` → `10` |
| Значение по умолчанию | `параметр=значение` | `def f(x, k=2): ...` | `k` можно не передавать |
| Минимум/максимум по признаку | `min(список, key=функция)` | `max(students, key=average)` | **элемент** с наибольшим значением признака |
| Короткая функция | `lambda аргумент: результат` | `lambda st: st.average()` | то же, что `def f(st): return st.average()` |

### 2. Новое: ООП

| Понятие | Что это | Как пишется | Пример |
|---|---|---|---|
| **Класс** | шаблон, чертёж | `class Имя:` (с большой буквы) | `class Student:` |
| **Объект** (экземпляр) | конкретная вещь по шаблону | `переменная = Класс(...)` | `a = Student("Аружан", "DA-01")` |
| **`__init__`** | вызывается **сам** при создании объекта, кладёт данные внутрь | `def __init__(self, ...):` | `def __init__(self, name, group):` |
| **`self`** | «этот объект»: тот, у которого вызвали метод | первый параметр каждого метода | `self.name = name` |
| **Атрибут** | данные внутри объекта | `объект.атрибут` — **без скобок** | `a.name`, `acc.balance` |
| **Метод** | функция внутри класса | `объект.метод()` — **со скобками** | `a.describe()`, `acc.deposit(1000)` |

**Шаблон класса:**
```python
class BankAccount:                          # 1. класс
    def __init__(self, owner, balance=0):   # 2. конструктор
        self.owner = owner                  # 3. атрибуты
        self.balance = balance

    def deposit(self, amount):              # 4. метод
        self.balance = self.balance + amount

acc = BankAccount("Аружан")                 # 5. объект
acc.deposit(1000)                           # 6. вызов метода
print(acc.balance)                          # 7. чтение атрибута → 1000
```

**Словарь → объект:**

| Словарь + функции | Класс |
|---|---|
| `student["name"]` | `a.name` |
| `average(student)` | `a.average()` |
| `add_grade(student, 90)` | `a.add_grade(90)` |
| словарь передаём в функцию | объект сам попадает в `self` |

### 3. Типичные ошибки

| Симптом | Причина | Как исправить |
|---|---|---|
| `<bound method ...>` вместо результата | забыли скобки у метода: `a.describe` | `a.describe()` |
| `if st.is_excellent == True:` никогда не срабатывает | без скобок это метод, а не `True`/`False` | `if st.is_excellent():` |
| `AttributeError: ... has no attribute 'name'` | в `__init__` написали `name = name` без `self` | `self.name = name` |
| `TypeError: __init__() missing 1 required positional argument` | при создании объекта передали не все данные | `Student("Мадина", "DA-02")` |
| `TypeError: 'float' object is not callable` | имя функции перезаписано числом: `min = 64.0` или `st.average = 91.7` | не называйте переменные `min`, `max`, `sum`, `list`, `print`; исправить: `del min` или Restart Kernel |
| Код «вдруг» сломался, хотя раньше работал | ячейки запускались вперемешку, в памяти старые объекты | Restart Kernel → запустить ячейки сверху вниз |

### 4. Почему функция и метод меняют объект «насовсем»
Переменная — это **имя**, которое указывает на объект. В функцию (и в `self`) попадает **тот же самый** объект, а не копия.

| Тип | Можно изменить на месте? | Изменение внутри функции видно снаружи? |
|---|---|---|
| `list`, `dict`, `set`, **объекты наших классов** | да | да: `append`, `d[key] = ...`, `self.balance = ...` |
| `int`, `float`, `str`, `tuple` | нет | нет |

> Правило: **метод/операция меняет объект** → видно снаружи. **`имя = новое`** внутри функции → создан новый объект, снаружи ничего не изменилось.

### Словарь — карточка с данными

````python
student = {"name": "Аружан", "group": "DA-01", "grades": [90, 75, 88]}

print(student["name"])          # достать значение по ключу
student["group"] = "DA-03"      # изменить значение
student["grades"].append(100)    # оценки — это список, у него есть append
print(student)
````

### Функция — действие над данными

````python
def average(student):
    return sum(student["grades"]) / len(student["grades"])

def add_grade(student, grade):
    student["grades"].append(grade)   

print(average(student))
add_grade(student, 100)
print(student["grades"])              
````

### Что здесь неудобно?
- Данные (`student`) и функции (`average`, `add_grade`) живут **отдельно**. Каждый раз передаём словарь в функцию.
- Можно опечататься в ключе: `student["grads"]` — и получить ошибку.
- Функцию можно вызвать с «чужим» словарём: `average({"price": 500})` — Python не остановит.

> **Класс** решает это: он **склеивает данные и функции в одно целое**. Функции внутри класса называются **методами** — и вы уже умеете их писать, это обычный `def`.

## 1. Классы и объекты

| Понятие | Что это | Пример |
|---|---|---|
| **Класс** | шаблон, чертёж | `Student` — «какие данные есть у любого студента» |
| **Объект** | конкретная вещь по шаблону | Аружан, Данияр, Мадина |
| **Атрибут** | данные внутри объекта | `a.name`, `a.group` — **без скобок** |
| **Метод** | функция внутри объекта | `a.describe()` — **со скобками** |

Вы уже пользовались объектами: `"алматы".upper()` — метод строки, `[1, 2].append(3)` — метод списка.

### Шаг 2. `__init__` и `self` — кладём данные внутрь
`__init__` вызывается **автоматически**, когда мы пишем `Student(...)`.
`self` — это **сам создаваемый объект**. `self.name = name` значит: «положи имя внутрь этого объекта».

````python
class Student:
    def __init__(self, name, group):
        self.name = name        # атрибут name
        self.group = group      # атрибут group

a = Student("Аружан", "DA-01")
b = Student("Данияр", "DA-01")

print(a.name)       # достаём атрибут через точку
print(b.name)

# class Teacher:
#     def __init__(self, n, g):
#         self.name = n
#         self.group = g

# a = Teacher("Thomas", "DA-01")
# print(a.name)
# print(a.group)
````

### Шаг 3. Первый метод
Метод — это `def` внутри класса. Первый параметр всегда `self`: через него метод видит данные **своего** объекта.

````python
class Student:
    def __init__(self, name, group):
        self.name = name
        self.group = group

    def describe(self):
        return f"{self.name}, группа {self.group}"

a = Student("Аружан", "DA-01")
b = Student("Данияр", "DA-01")

print(a.describe())     # один и тот же метод...
print(b.describe())     # ...но у каждого объекта свои данные
````

> `a.describe()` — это то же самое, что `Student.describe(a)`. Python сам подставляет объект на место `self`. Помните разминку: `self` играет ту же роль, что словарь `student`, который мы передавали в функцию.

### 2.1. `Student` с оценками
Добавим студенту список оценок и методы, которые с ними работают. Это та же логика, что в повторении, — но теперь данные и функции **внутри одного класса**.

````python
class Student:
    def __init__(self, name, group):
        self.name = name
        self.group = group
        self.grades = []                 # у нового студента оценок нет

    def add_grade(self, grade):
        if 0 <= grade <= 100:
            self.grades.append(grade)
        else:
            print(f"Оценка {grade} вне диапазона 0–100")

    def average(self):
        if len(self.grades) == 0:
            return 0
        return round(sum(self.grades) / len(self.grades), 1)

    def is_excellent(self):
        return self.average() >= 90      # метод вызывает другой метод через self
````

````python
aruzhan = Student("Аружан", "DA-01")
aruzhan.add_grade(92)
aruzhan.add_grade(88)
aruzhan.add_grade(105)       # ПРЕДСКАЖИТЕ: что произойдёт?

print(aruzhan.grades)
print(aruzhan.average())
print(aruzhan.is_excellent())
````

**Сравните с повторением:**

| Было (словарь + функции) | Стало (класс) |
|---|---|
| `add_grade(student, 92)` | `aruzhan.add_grade(92)` |
| `average(student)` | `aruzhan.average()` |
| проверку диапазона легко забыть | проверка живёт **внутри** метода |

### 2.2. `BankAccount` — счёт с проверками
Баланс меняется только через методы, а методы проверяют данные.

````python
class BankAccount:
    def __init__(self, owner, balance=0):     # balance по умолчанию 0
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Сумма должна быть больше нуля")
        else:
            self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Недостаточно средств: на счёте {self.balance} ₸")
        else:
            self.balance = self.balance - amount

    def show_balance(self):
        print(f"{self.owner}: {self.balance} ₸")
````

````python
acc = BankAccount("Данияр")
acc.deposit(450_000)        # зарплата
acc.withdraw(180_000)       # аренда
acc.show_balance()
acc.withdraw(1_000_000)     # ПРЕДСКАЖИТЕ: что напечатается и каким станет баланс?
acc.show_balance()
````

````python
# ПРЕДСКАЖИТЕ: что выведется? Чем отличаются эти две строки?
print(acc.balance)
acc.show_balance()
````
