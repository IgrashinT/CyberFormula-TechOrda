# Урок 8 · Челленджи

Три мини-проекта посложнее ДЗ, в каждом по 3 шага: **1** — журнал оценок группы, **2** — анализ двух текстов, **3** — мини-магазин. Шаги идут по порядку, и каждый использует предыдущий.

**Как работать.** Сначала запустите ячейку с функцией `check`. В каждой задаче замените `...` на свой код и запустите ячейку: ✅ — верно, ❌ — подсказка, что получилось не так. Не страшно, если что-то не выходит с первого раза — перечитайте напоминание вверху или тетрадь урока 8.

```python
# ▶ Запустите эту ячейку первой — она нужна для проверки задач
def check(name, func, expected):
    try:
        got = func()
    except Exception as e:
        print(f"❌ {name}: ошибка {type(e).__name__}: {e}")
        return
    if got == expected:
        print(f"✅ {name}")
    else:
        print(f"❌ {name}: получилось {got!r}, а нужно {expected!r}")

print("Готово!")
```
