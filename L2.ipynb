## Часть 3. Переменные, типы данных и вычисления — 30 минут

### 3.1. Переменная — это имя для значения
`age = 18` означает «сохранить значение 18 под именем age». Знак `=` здесь **присваивает** значение, а не сравнивает.

```python
age = 18
city = "Астана"
print(age)
print(city)
age = age + 1
print("Через год:", age)
```

### 3.2. Четыре основных типа
- `int` — целое число: `20`
- `float` — число с дробной частью: `3.5` (точка, не запятая)
- `str` — текст: `"Python"`
- `bool` — логическое значение: `True` или `False`

`type()` показывает тип.

```python
students = 12
price = 29.5
course = "Python"
is_beginner = True
print(type(students), type(price), type(course), type(is_beginner))
```

### 3.3. Арифметические операции
`+` сложить · `-` вычесть · `*` умножить · `/` разделить · `//` целая часть деления · `%` остаток · `**` степень. Не делите на ноль.

```python
a = 17
b = 5
print("Сумма:", a + b)
print("Разность:", a - b)
print("Произведение:", a * b)
print("Деление:", a / b)
print("Целая часть:", a // b)
print("Остаток:", a % b)
print("Степень:", a ** 2)
```

### 3.4. Преобразование типов
Если введено `"12"`, это текст. Для арифметики используем `int()` или `float()`. В этом уроке предполагаем, что пользователь вводит правильное число. Обработку ошибок пройдём позже.

```python
first = float(input("Введите первое число (например, 4.5): "))
second = float(input("Введите второе число (например, 2): "))
print("Их сумма:", first + second)
```

### 3.5. f-строки — удобное оформление текста
Поставьте `f` перед кавычками; в фигурных скобках можно писать имена переменных и выражения.

```python
product = "тетрадь"
count = 3
unit_price = 450
print(f"{count} шт. × {unit_price} ₸ = {count * unit_price} ₸ за {product}")
```

```python
product = "тетрадь"
count = 3
unit_price = 450

# Вариант через конкатенацию (сложение строк)
print(str(count) + " шт. × " + str(unit_price) + " ₸ = " + str(count * unit_price) + " ₸ за " + product)
```

### Challenge 1. Чек интернет-магазина

Пользователь вводит **название товара**, **цену одной единицы** (может быть дробной), **количество**, **скидку в процентах** и **стоимость доставки**.

Выведите аккуратный чек с помощью **f-строк**: товар, количество, исходную стоимость, сумму скидки, цену после скидки, доставку и итог. Все денежные суммы покажите с **2 знаками после точки**.

Формулы:
- Стоимость = цена × количество
- Скидка в деньгах = стоимость × процент скидки / 100
- Итого = стоимость − скидка в деньгах + доставка

**Пример:** товар «Книга», цена 2500.50, количество 3, скидка 10%, доставка 600 → **7351.35**.

**Ограничение:** ввод через `input()`, используйте `float()` и `int()`; без условий, циклов, функций.


```python
# Challenge 1 — создайте чек интернет-магазина
# 1. Считайте данные через input()
nam = input("Название товара: ")
cst_one = float(input("Цена одной единицы: "))
num = int(input("Количество: "))
discnt = float(input("Скидка в проентах: "))
cst_delvr = float(input("Стоимость доставки: "))

# 2. Вычислите суммы
cst_ord = cst_one*num
cst_final = cst_ord - cst_ord*(discnt/100) + cst_delvr

# 3. Выведите чек с помощью f-строк
print(f"товар «{nam}», цена {cst_one:.2f}, количество {num}, скидка {discnt}%, доставка {cst_delvr:.2f} → {cst_final:.2f}.")
```

### Challenge 2. Калькулятор времени выполнения задач

Пользователь вводит общее количество минут, потраченных на работу (целое число), а также стоимость **одного часа** работы (дробное число).

Выведите:
1. Полное число часов и оставшиеся минуты, используя `//` и `%`.
2. Время в часах в виде десятичного числа с **2 знаками после точки**.
3. Стоимость работы по фактическому времени с **2 знаками после точки**.

**Пример:** 135 минут и 2400 за час → `2 ч 15 мин`, `2.25 часа`, `5400.00`.

**Дополнительная проверка:** код должен также правильно работать для 45 и 185 минут без каких-либо изменений.

**Ограничение:** без условий, циклов и функций.


```python
# Challenge 2 — рассчитайте часы, минуты и стоимость
mins = int(input("Количество минут: "))
cst_one = float(input("Стоимость одного часа: "))

hr_int = mins // 60
min_int = mins % 60
hr_whl = mins / 60
cst_all = cst_one * hr_whl

print("{} минут и {} за час → {} ч {} мин, {:.2f} часа, {:.2f}.".format(mins, cst_one, hr_int, min_int, hr_whl, cst_all))
```

### Challenge 3. Финансовый мини-отчёт с типами данных

Пользователь вводит **месячный доход**, **расходы на жильё**, **расходы на питание** и **другие расходы** (все суммы могут быть дробными).

Рассчитайте и выведите через f-строки:
1. Общие расходы.
2. Остаток средств.
3. Долю каждой категории расходов от дохода в процентах с точностью **1 знак после точки**.
4. Долю остатка от дохода в процентах.
5. Результат `type()` для дохода и текстового названия отчёта.

**Пример для проверки:** доход 200000, жильё 60000, питание 40000, другое 20000 → расходы 120000, остаток 80000, доли 30.0%, 20.0%, 10.0%, остаток 40.0%.
